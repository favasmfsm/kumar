```python
import os
import glob

folder_path = "../processed_data/"

df1 = pd.DataFrame()

# Find all Parquet files in the folder
parquet_files = glob.glob(os.path.join(folder_path, "*.parquet"))

for file in parquet_files:
    df = pd.read_parquet(file)
    # df['date'] = pd.to_datetime(df['date'])
    # df['year'] = df['date'].dt.year
    combined = df[df.cik.isin(breach_list.cik.astype(str))]
    # Determine if full_content is needed
    # Keep full_content only if item_7 or risk_factors is NA
    # needs_full_content = df['item_7'].isna() & df['risk_factors'].isna()

    # Separate the rows based on condition
    # df_keep_full = df.loc[needs_full_content, ['cik', 'date', 'file', 'full_content', 'item_7', 'risk_factors', 'item_1a_flag', 'item_7_flag']]
    # df_drop_full = df.loc[~needs_full_content, ['cik', 'date', 'file', 'item_7', 'risk_factors', 'item_1a_flag', 'item_7_flag']]

    # # Add a full_content column with None to match the schema for concat
    # df_drop_full['full_content'] = None

    # # Combine the two parts and concat to df1
    # combined = pd.concat([df_keep_full, df_drop_full], ignore_index=True)
    df1 = pd.concat([df1, combined], ignore_index=True)

    print(f"Processed: {file}")

    
# Read and concatenate them into a single DataFrame
# df = pd.concat([pd.read_parquet(file) for file in parquet_files], ignore_index=True)
```
